## looking at the person table several things are apparent:
  - data is de-identified (dates like year of birth are shifted)
  - you mainly see two types of columns:  source values & concepts
  - source values in OMOP are typically the values from columns in source datasets prior to OMOP mapping

    
- concepts are the mapped OMOP values -> OMOP functions by mapping source values onto the SNOMED ontology, hence all source values should have corresponding SNOMED concepts in every table; this is done to facilitate OMOP working off 1 single vocabulary

  
- you can access the OMOP concept table for more information about individual concepts, such as the descriptions of the concept id's
